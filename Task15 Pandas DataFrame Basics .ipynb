```python

import pandas as pd
from sklearn.datasets import load_iris

# 1. Create DataFrame manually
data = {
    'Name': ['Ravi', 'Sita', 'Ram', 'Priya'],
    'Age': [20, 21, 19, 22],
    'Marks': [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print("Manual DataFrame:")
print(df)

# 2. Display first rows
print("\nFirst Rows:")
print(df.head())

# 3. Display shape
print("\nShape:")
print(df.shape)

# 4. Display columns
print("\nColumns:")
print(df.columns)

# 5. Display information
print("\nData Information:")
df.info()

# 6. Display summary statistics
print("\nSummary Statistics:")
print(df.describe())

# 7. Load real Iris dataset
iris = load_iris()

iris_df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

print("\nIris Dataset:")
print(iris_df.head())

print("\nIris Dataset Shape:")
print(iris_df.shape)

print("\nIris Dataset Summary:")
print(iris_df.describe())
```
